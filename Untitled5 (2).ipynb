```python
# indexing and slicing
A = np.array([[1,2,3],[4,5,6]])
print(A[0,0])   # burada indexleme aynı listelerde olduğu gibi ilk eelman 0 dan başlıyor
print(A[-1,-1]) # iki indexe de tersten başlayarak bakılıyor, sol satırı sağ da sütunu ifade eder. # buraya kadar indexing

# slicing                    # bu konuda bir şey yazmıyorsa ifade ettiği satır veya sutünun tamamı demektir
#slicing in vectors
V = np.array([1,2,3,4,5,6])
print(V[ : 2]) # 0. indexten başlayıp 3. ye kadar yazdı ama 3. dahil değil
print(V[ : : 2]) # tüm vectorde 2şer 2 şer atlayarak
print(V[::-1])  # tersten başlayarak zılayarak gider burada bir zıplıyor altta 2
print(V[::-2])

# slicing in matris  Matrislerde virgül , sol tarafı satırlar, sağ tarafı sütunlar için ayırır.
m = np.array([[1,1,2],[2,3,4],[3,2,3]])
print(m[ 1, :1])  # burada indexlemede 2.satırı alıyoruz tüm elemanlarından 0. dan 1. e kadar yani çıktımız 2 olacak
print(m[ :2,1])  # bu 1. satırdan 3 satıra kadar (3. dahil değil) bunun 2. sütununu alır
print(m[:1,:2])  # burada 1. satırdan 2. ye , 1. sutündan 3. sütuna kadar alır . ve submatrix oluşturmuş oluruz
```

```python
import numpy as np
print("--- SENARYO 1: Skaler (0D) ve Matris (2D) ---")
# Skaler sayı, matrisin HER BİR elemanına sanal olarak dağıtılır.

matris_2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])  # Shape: (2, 3)

skaler = 10  # Shape: () -> Tek sayı

# Arka planda skaler 10 sayısı [[10, 10, 10], [10, 10, 10]] şeklini alır.
sonuc_1 = matris_2d + skaler
print("Matris + Skaler:\n", sonuc_1)
print("-" * 50)


print("--- SENARYO 2: Matris (2D) ve Vektör (1D) [Satır Bazlı Ekleme] ---")
# En sık kullanılan senaryodur (Örn: Veriden ortalama çıkarma).

A = np.array([
    [10, 20, 30],
    [40, 50, 60]
])  # Shape: (2, 3) -> 2 satır, 3 sütun

v_satir = np.array([1, 2, 3])  # Shape: (3,) -> NumPy bunu (1, 3) gibi görür.

# BOYUT KONTROLÜ (Sağdan Sola):
# A shape:       (2, 3)
# v_satir shape: (   3)  --> Sağdaki 3 == 3 (Eşit!), Soldaki 2 ve 1 (1 esnetilir)
# Arka planda v_satir, [[1, 2, 3], [1, 2, 3]] şekline bürünür.

sonuc_2 = A + v_satir
print("Matris + Satır Vektörü:\n", sonuc_2)
print("-" * 50)


print("--- SENARYO 3: Sütun Vektörü (2D) ve Satır Vektörü (1D) ---")
# İki tarafın da bir boyutu 1 olduğunda ikisi de birbirine doğru genişler ve 2D matris üretir.

sutun_vektoru = np.array([
    [100],
    [200]
])  # Shape: (2, 1) -> 2 satır, 1 sütun

satir_vektoru = np.array([1, 2, 3])  # Shape: (1, 3)

# BOYUT KONTROLÜ:
# sutun_vektoru: (2, 1)  --> 1 boyutu 3'e esnetilir (sütunlar çoğalır)
# satir_vektoru: (1, 3)  --> 1 boyutu 2'ye esnetilir (satırlar çoğalır)
# Sonuç Shape: (2, 3) olur!

sonuc_3 = sutun_vektoru + satir_vektoru
print("Sütun Vektörü + Satır Vektörü:\n", sonuc_3)
print("-" * 50)


print("--- SENARYO 4: Hata Durumu ve `.reshape()` ile Çözümü ---")

# Hatalı Durum Örneği:
# matris_2d shape: (2, 3)
# hatali_vektor:   (2,)   --> Sağdan kontrol: 3 ve 2 EŞİT DEĞİL ve BİRİ 1 DEĞİL!
# matris_2d + np.array([10, 20])  # <-- Bu satır ValueError hatası fırlatır!

# ÇÖZÜM: Vektörü Sütun Vektörüne (2, 1) Dönüştürmek!
duzgun_vektor = np.array([10, 20]).reshape(2, 1)  # Shape: (2, 1)

# BOYUT KONTROLÜ:
# matris_2d:     (2, 3)
# duzgun_vektor: (2, 1)  --> En sağdaki 1 boyutu 3'e esnetilir. İŞLEM BAŞARILI!

sonuc_4 = matris_2d + duzgun_vektor
print("Sütun Bazlı Ekleme Sonucu:\n", sonuc_4)
```

```python
# BOOLEAN MASK
# Bir NumPy dizisine mantıksal bir karşılaştırma (>, <, ==, !=) uyguladığında, NumPy sana sayılardan oluşan değil,
# True ve False değerlerinden oluşan aynı boyutta yeni bir dizi döndürür. Buna maske denir.
import numpy as np
vector = np.array([1,2,3,4,5])
mask = vector > 2
print(mask)
req_vector = vector[mask]
print(req_vector)
```

```python
#np.where ile değişken değiştirme
sicakliklar = np.array([-5, 12, 0, 28, -2, 35])

# Sıcaklık 0'dan küçükse "Dondurucu", değilse "Normal" diyelim:
durum = np.where(sicakliklar < 0, "Dondurucu", "Normal")
print(durum)
# Çıktı: ['Dondurucu' 'Normal' 'Normal' 'Normal' 'Dondurucu' 'Normal']

# Sıfırın altındaki değerleri 0'a eşitleyelim (Clipping / Ayıklama):
temiz_sicaklik = np.where(sicakliklar < 0, 0, sicakliklar)
print(temiz_sicaklik)
# Çıktı: [ 0 12  0 28  0 35]
```

```python
import numpy as np

# =============================================================================
# NUMPY AŞAMA 4: BOOLEAN INDEXING & MASKELEME REHBERİ
# =============================================================================

# 1. 2D MATRİSTE MASKELEME
matris = np.array([
    [12, 45, 78],
    [34, 89, 23],
    [67, 11, 90]
])

print("Orijinal Matris:\n", matris)

# 50'den büyük elemanları filtreleme
# NOT: 2D matrisi maskelediğinde sonuç her zaman 1D (tek sıra) vektör döner!
buyukler = matris[matris > 50]
print("\n50'den Büyük Elemanlar (1D Vektör):", buyukler)


# 2. MASKE İLE DEĞER ATAMA (In-place Mutation)
# Matris içindeki 50'den küçük TÜM elemanları doğrudan 0 yapalım:
matris_kopya = matris.copy()
matris_kopya[matris_kopya < 50] = 0

print("\n50'den Küçükleri 0 Yapılmış Matris:\n", matris_kopya)


# 3. NP.WHERE İLE KOŞULLU DÖNÜŞTÜRME (IF-ELSE)
# Çift sayılara 1, tek sayılara -1 yazalım:
isaretli_matris = np.where(matris % 2 == 0, 1, -1)
print("\nÇiftler 1, Tekler -1 Yapılmış Matris:\n", isaretli_matris)
```

```python
import numpy as np
temp = np.random.seed(1)   # bu satır phytonun hep aynı veri setini vermesini sağlıyor nasıl hesaspladığını bilmen gerekmiyor
print("new data creation:")
temp = np.random.uniform(20,35,20)
temp = np.round(temp,1)  # round komutu virgülden sonraki kacıncı haneye yuvarlayacağımızı yapıyor bu komut olmasaydı çok uzun virgüllü sayılar gelir
print(temp)
print("adding noise")
temp[12] = -99
temp[5]= -99
print("*"*100)
print("slicing and reshape")
first3_day = temp[:12]
print("first 3 days temparatures", np.round(first3_day,1))   # tekrar round çalıştırdık çünkü olmasaydı tekrar uzun ondalıklı sayı verecekti
print("*"*100)
temp_matrix = temp.reshape(5,4)
print(np.round(temp_matrix,1))  # tekrar round la yazdık
print("*"*100)
print("broadcasting and boolean indexing")
offset = np.array([0.1,-0.3,2.0,-1.5])
calibrated_matrix = temp_matrix + offset
print("calibrated matrix: " , calibrated_matrix) # bu offsetle toplamayı broadcastingi yapabildiğim için yazdım bir işlevi yoktur.
error_mask = calibrated_matrix < 0
print("missed datas:", error_mask )
clean_matrix = np.where(calibrated_matrix<0,25,calibrated_matrix)
print("clean matrix", np.round(clean_matrix,1))
high_temp = clean_matrix[clean_matrix > 30]
print("high temperatures: ", np.round(high_temp,1))
print("*"*100)
print("analysis, axis, numeric")
dailymeans = np.mean(clean_matrix, axis = 1)
print("mean teöparature values:", np.round(dailymeans,1))
maxtemp = np.max(clean_matrix)
print("max temparature: ", np.round(maxtemp,1))
```

```python
import numpy as np
notes = np.random.seed(42)
print("new notes creation")
notes = np.random.uniform(10,100,20)
notes = np.round(notes,1)
print(notes)
print("*"*100)
print("reshape the array ")
notes_matrix = notes.reshape(5,4)
print(notes_matrix)
print("adding noise")
notes_matrix[0,1]= -10
notes_matrix[3,2]= -5
print(notes_matrix)
print("broadcasting (adding bonus)")
bonus = np.array([5, 10, 0, 15])
real_notes = notes_matrix + bonus
print(real_notes)
print("masking and boolean indexing")
error_mask = real_notes<0
print(error_mask)
clean_notes = np.where(real_notes<0,0,real_notes)
print(clean_notes)
high_notes = clean_notes[clean_notes>=70]     #buradaki kodumuz şöyle de olabilirdi mask = clean_notes>=70
np.round(clean_notes,1)
print(high_notes)                             #                                      high_notes = clean_note[mask]
print("analysing")
student_avg = np.mean(clean_notes, axis = 1) #axis=1 olduğunda her öğrencinin ders nnotları ortalamsı (satır gelir),
print(student_avg)
course_avg = np.mean(clean_notes,axis = 0)   #axis = 0 olduğunda her dersin ortalamarı (sütun gelir)
print(course_avg)
course_max = np.max(clean_notes, axis = 0)
print(course_max)
student_max =np.max(clean_notes, axis = 1)
print(student_max)
course_max_index = np.argmax(clean_notes, axis = 0)
print(course_max_index)
```
