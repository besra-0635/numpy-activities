```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data = {
    'ay' : ['ocak','şubat','mart','nisan'],
    'satiş': [1000,2000,3000,4000],
    'kategori': ['yiyecek','yiyecek','giyim','giyim']
}
df = pd.DataFrame(data)
plt.figure(figsize=(8,4))
sns.lineplot( data = df , x= 'ay', y= 'satiş', marker = 'o')
plt.title('aylara göre satış grafiği')
plt.xlabel('ay')
plt.ylabel('getiri')
plt.show()
```

```python
import pandas as pd    #sns.lineplot çizgi grafiği
import matplotlib.pyplot as plt
import seaborn as sns
data = {
  'gün' : ['pzt','salı','çarş','perş','cuma'],
  'sıcaklık':[22, 25, 21, 28, 26]
}


df = pd.DataFrame(data)
print(df)
plt.figure(figsize=(10,5))
sns.lineplot(data= df, x= 'gün',y= 'sıcaklık', marker = '*')  # marker neyse peak noktalara onu koyar
plt.title('haftaiçi sıcaklık değerleri')
plt.xlabel('günler')
plt.ylabel('sıcaklık')
plt.show()
```

```python
# sütun grafiği
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data = {
  'gün' : ['pzt','salı','çarş','perş','cuma'],
  'sıcaklık':[22, 25, 21, 28, 26]
}
df = pd.DataFrame(data)
print(df)
plt.figure(figsize=(10,5))
sns.barplot(data = df , x = 'gün', y = 'sıcaklık', palette='viridis')  # palette ifadesi hangi renkleri kullanacağını belirtiyor.
plt.title('haftaiçi sıcaklık değerleri')
plt.show()
```

```python
# scatterplot
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
veri = {
    'metrekare':[80,90,70,50,60,70],
    'fiyat':[49576,28943,57656,86765,54644,45644]  # bu grafik türünde eğer iki şey arasında bir yönelim bir ilişki yoksa (randomly) dağıldıysa bu veriyi modellemede kullanamam.
}                                                  # ama bir ilişki bulabilirsek kullanmak mantıklı.
df = pd.DataFrame(veri)
plt.figure(figsize = (8,8))
sns.scatterplot(data = df , x='metrekare',y= 'fiyat')
plt.xlabel('ev metrekaresi')
plt.ylabel('fiyat')
plt.title('metrekareyle fiyat ilişkisi')
plt.show()
```

```python
# histogram (sns.histplot)
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
veri = {
    'metrekare':[80,90,70,50,60,70],
    'fiyat':[49576,28943,57656,86765,54644,45644]
}
df = pd.DataFrame(veri)
plt.figure(figsize=(4,4))
sns.histplot(data = df, x = 'metrekare', bins= 4, kde= True , color= 'green')
plt.show()
```

```python
# heatmap correlasion
#İki sayısal sütun arasındaki ilişkinin -1 ile +1 arasındaki sayısal değeridir:

#+1'e yakınsa: Birlikte artıyorlar (örn: Metrekare arttıkça Fiyat artar).

#-1'e yakınsa: Biri artarken diğeri azalıyor (örn: Yaş arttıkça Fiyat düşer).

#0'a yakınsa: Aralarında ilişki yoktur.
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
info = {
    'metrekare': [60,70,80,90,100],
    'oda sayısı':[2,3,4,5,6],
    'fiyat': [1.5, 2.3, 3.1, 3.8, 4.9]
    }
df = pd.DataFrame(info)
correlasion = df.corr()
plt.figure(figsize= (5,5))
sns.heatmap(data=correlasion , annot = True, cmap= 'coolwarm' )
plt.title('sütunlar arası korrelasyon')
plt.show()
```
